Course Workflow Migration Plan
From Jupyter Lab to Google Colab

Technical & Operational Assessment for Local-to-Cloud Workflow Migration  

- What is Untouched

Notebook Files will be untouched keeping the notebook structure, ordering, and flow without modifying them to run. Core programing codes and packages that will be used like pandas will be the same when used with using the python language. In additon, inside the coding the markdown and comments made will also remain untouched since those are records of changes made to in order to analyze the data and the reasoning behind it.

- What Needs Redoing

When moving from local hardware to a managed cloud VM like colabs, will require possible environment setup, file handling, and repository syntax to match.

Since programs that have been done on my machine, the transistion to the cloud machine will need login access like that in colab to sign onto the services for the virtual machine and google drive access. Since code in the computer is stored locally, there is no access for others to have access to it, unless its loaded onto Github or other means of access physically.  While in colab, different permissions and access will need to be granted in order for the programs to be run or have access to.

- Experience-Grounded Observations 

Noticed there are times when being idle from Colab leads disconnections that wipe any temporary files you uploaded or coding since rented time means there a time limit to having the servers up and running. Moving from a permanent local hard drive to Colab's temporary storage will be a major adjustment. In addition, installing a package on a local computer lasts forever where in Colab, the virtual machine resets every session, requiring you to re-run your install evertime.

- Key Risk & Mitigation

Risks
One of the risks for the migration would be when moving to Colab which is tied to a remote Google servers over the internet. If Wi-Fi or internet drops unexecuted code stops immediately and unsaved information can become lost. Also since the computer is rented time, there are cases here free-tier runtimes also enforce strict usage limits due to the "free" prcessig power that is being offered. Since it was mentioned that if you ask for too mcuh you'll never utilize all the machine being paid for, while on the other hand if using less than whats needed, it'll become a bottleneck.

Mitigation Strategy
Save intermediate outputs directly to mounted Google Drive storage so there is no loss of information if communication is dropped. Then importat to have a backup and push updated copies frequently to avoid information loss and redundancy for code in case one set is compromised.
