```python
#Null Hypothesis - There is no significant difference in weight between sunflower fed chicks and horsebean fed chicks.
```

```python
df=pd.read_csv("Downloads/chickwts.csv")
from scipy.stats import ttest_ind
a = df[df["feed"] == "sunflower"]["weight"]
b = df[df["feed"] == "horsebean"]["weight"]
ttest_ind(a, b)
```

```python
#Chicks fed sunflower meal weighed an average of 168.7 grams more than those fed horsebean the p value being 2.3743507676837164e-08, leading to the rejection of the null hypothesis,this finding should be used with caution given the sample size.
```
