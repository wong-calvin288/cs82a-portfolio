```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.isnull())
```

```python
housing_clean=df.dropna(subset=["housing.type"])
```

```python
housing_clean.isnull().sum()
```

```python
housing_counts = (housing_clean["housing.type"].value_counts().reset_index())
housing_counts.columns = ["housing_type", "count"]
plt.figure(figsize=(10,5))
sns.barplot(data=housing_counts,x="housing_type",y="count")
plt.title("Homeowners with mortgage and renters making up the majority of customers")
plt.xlabel("Housing Types")
plt.ylabel("Count")
plt.show()
```
