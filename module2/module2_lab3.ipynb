```python
#temp cut off range in fahrenheit
#freezing = 32 and below
#cold = 33 to 70
#mild = 71 to 80
#warm = 81 to 90
#hot = 90 and above
Temperatures =[32,21,31,32,33,69,70,71,79,80,81,88,89,90,91]
for t in Temperatures:
    if (t<=32):
        print (f"{t} Fahrenheit is Freezing")
    elif (t >32) and (t<=70):
        print (f"{t} Fahrenheit is Cold")
    elif (t>70) and (t<=80):
        print (f"{t} Fahrenheit is Mild")
    elif (t >80) and (t <=90):
        print (f"{t} Fahrenheit is Warm")
    elif (t >90):
        print (f"{t} Fahrenheit is Hot")
    else:
        print()
```
