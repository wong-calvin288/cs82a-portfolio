```python
def adding_total_if(a,b):
    b_negative=False
    if  b<0:
        b=-b
        b_negative=True
    total = 0
    for c in range(b):
        total = total +a
    if b_negative:
        total=-total
    return total
print(adding_total_if(-7,-6))
```

```python
def adding_total_while(x,y):
    y_negative=False
    if  y<0:
        y=-y
        y_negative=True
    total = 0
    while y>0:
        total = total+x
        y=y-1
    if y_negative:
        total=-total
    return total
print(adding_total_while(7,6))
```
