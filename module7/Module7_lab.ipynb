```python
#step 1 Load tips.csv and frame the problem
```

```python
import pandas as pd
```

```python
import sklearn
```

```python
df=pd.read_csv('Downloads/tips.csv')
```

```python
df.head
```

```python
df.describe()
```

```python
from sklearn.model_selection import train_test_split
X = df[["total_bill"]]
y = df["tip"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)

print(f"len(X_train): {len(X_train)}")
```

```python
# step 2 Why hold out 30%? (markdown cell)
```

By hold out 30% of our data as an unseen test set, it will test to see how well our model will work with new data rather than using data it was trained on. Testing on unseen data prevents overfitting and gives a better measure of the models training.

```python
# step 3  Fit the regression and read the slope
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
```

```python
print(model.coef_, model.intercept_)
```

each extra dollar of bill adds about .11 of tip.

```python
#step 4 Draw the fitted line
```

```python
import matplotlib.pyplot as plt
```

```python
y_predict=model.predict(X)
```

```python
plt.figure(figsize=(12, 8))
plt.scatter(X, y)
plt.plot(X, y_predict)


plt.xlabel('total_bill')
plt.ylabel('tip')
plt.title('Total Bill vs. Tip with Fitted Line')

plt.show()
```

```python
#step 5 load iris and split it
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
iris = load_iris(as_frame=True)
X_iris = iris.data
y_iris = iris.target
```

```python
X_train_i, X_test_i, y_train_i, y_test_i = train_test_split(
    X_iris, y_iris, test_size=0.3, random_state=42)
```

```python
print(f"Total flowers: {len(X_iris)}")
print(f"Training flowers: {len(X_train_i)}")
```

```python
#step 6 fit and score kNN
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
knn = KNeighborsClassifier(n_neighbors=5)
tree = DecisionTreeClassifier(max_depth=3)
knn.fit(X_train_i, y_train_i)
print(knn.score(X_test_i, y_test_i))
```

```python
# step 7 Fit, score, and read a shallow tree
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
tree.fit(X_train_i, y_train_i)
print(f"Test score: {tree.score(X_test_i, y_test_i):.2f}")
```

```python
plt.figure(figsize=(10, 6))
plot_tree(tree, feature_names=X_train_i.columns,
        class_names=[
        "setosa",
        "versicolor",
        "virginica",]
          ,filled=True)
plt.show()
```

Tree Diagram Meaning - The decision tree first checks whether the flower's petal length is 0.8 cm or less. If true, the flower is immediately classified as a setosa; if not, the tree moves down to evaluate other features for the remaining types.

```python
#Step 8: The overfitting probe
```

```python
from sklearn.tree import DecisionTreeClassifier
tree_deep = DecisionTreeClassifier(random_state=42)
tree_deep.fit(X_train_i, y_train_i)
print(f"{'Model':<25} | {'Train Score':<12} | {'Test Score':<12}")
print("-" * 55)
print(f"{'Shallow Tree (depth=3)':<25} | {tree.score(X_train_i, y_train_i):<12.2f} | {tree.score(X_test_i, y_test_i):<12.2f}")
print(f"{'Deep Tree (no limit)':<25} | {tree_deep.score(X_train_i, y_train_i):<12.2f} | {tree_deep.score(X_test_i, y_test_i):<12.2f}")
```


* Shallow Tree at max_depth=3 shows a slight performance gap between training 0.95 and testing 1.00 scores. Meaning it'll handle a new test.

* Deep Tree at max_depth=None shows at no limit is able to get 1.00 on the training score by memorizing the training data, showing an overfitting compared to the limited model.

```python
# Step 9 - Predict one new flower
```

```python
new_flower = [[5.1, 3.5, 1.4, 0.2]]
```

```python
prediction = tree_deep.predict(new_flower)
```

```python
predicted_name = iris.target_names[prediction[0]]
print(f"Predicted Class: {predicted_name}")
```

```python
# Step 10: Conclusion & Summary
```


* Model Performance - Both kNN k=5 and the shallow Decision Tree of max_depth=3 shows the program was able to look at the data and tell the flowers apart

* Overfitting & Generalization - When Decision Tree max_depth=None achieved a 1.00 training score because of memorization.  Since it's memorized, if new information were tested there will be issues.  Thats why there needs to be limits and rules to allow for learning.

* Interpretability - Decision trees show a flowchart thats easy to follow by comparing observed values and using the charts to easily show the logic

* Failure - This model might fail if presented with rare flower with measurments outside the range or have measurement errors

* Human Application - If used on people instead of flowers, it could easily turn into a tool for stereotypes. Because the decision trees make strict yes and no or true and false based on data, the people can end up being catergorized into the wrong categories
